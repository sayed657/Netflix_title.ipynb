```python
df['director'].fillna('Unknown',inplace=True)
df['country'].fillna('Unknown',inplace=True)
df['cast'].fillna('Unknown',inplace=True)

```

```python
x = df['rating'].mode()
df['rating'].fillna(x[0],inplace=True)
```

```python
x = df['date_added'].mode()
df['date_added'].fillna(x[0],inplace=True)
```

```python
x = df['duration'].mode()
df['duration'].fillna(x[0],inplace=True)
```
